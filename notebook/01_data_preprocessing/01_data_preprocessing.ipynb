# Step 1 — Data Preprocessing

Load raw data, inspect quality, handle missing values, encode target, and save cleaned data.

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../../data/raw/german_credit_data.csv', index_col=0)
print('Shape:', df.shape)
df.head()
```

```python
# Basic info
print('=== Data Types ===')
print(df.dtypes)
print('\n=== Missing Values ===')
print(df.isnull().sum())
print('\n=== Duplicates ===')
print('Duplicate rows:', df.duplicated().sum())
print('\n=== Class Distribution ===')
print(df['Risk'].value_counts())
```

```python
# Statistical summary
df.describe(include='all')
```

```python
# Fill missing categorical columns with mode
for col in ['Saving accounts', 'Checking account']:
    mode_val = df[col].mode()[0]
    df[col].fillna(mode_val, inplace=True)
    print(f'{col}: filled with "{mode_val}"')

# Encode target: good=0, bad=1
df['Risk'] = df['Risk'].map({'good': 0, 'bad': 1})

print('\nMissing after fill:', df.isnull().sum().sum())
print('Target encoded:', df['Risk'].value_counts().to_dict())
```

```python
# Save cleaned data
df.to_csv('../../data/processed/cleaned_data.csv')
print('Saved: data/processed/cleaned_data.csv')
df.head()
```
