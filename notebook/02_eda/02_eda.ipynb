# Step 2 — Exploratory Data Analysis (EDA)

Visualize distributions, class balance, relationships between features and target.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../../data/processed/cleaned_data.csv', index_col=0)
print('Shape:', df.shape)
df.head()
```

```python
# Target class distribution
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df['Risk'].value_counts().plot(kind='bar', ax=axes[0], color=['steelblue', 'tomato'], edgecolor='black')
axes[0].set_title('Class Distribution (0=Good, 1=Bad)')
axes[0].set_xticklabels(['Good (0)', 'Bad (1)'], rotation=0)

df['Risk'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%',
                                colors=['steelblue', 'tomato'], labels=['Good', 'Bad'])
axes[1].set_ylabel('')
axes[1].set_title('Class Balance')
plt.tight_layout()
plt.show()
```

```python
# Numeric feature distributions
num_cols = ['Age', 'Credit amount', 'Duration']
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, num_cols):
    df[col].hist(ax=ax, bins=30, edgecolor='black', color='steelblue')
    ax.set_title(f'Distribution: {col}')
    ax.set_xlabel(col)
plt.suptitle('Numeric Feature Distributions')
plt.tight_layout()
plt.show()
```

```python
# Numeric features by Risk (box plots)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, num_cols):
    df.boxplot(column=col, by='Risk', ax=ax)
    ax.set_title(f'{col} by Risk')
    ax.set_xlabel('Risk (0=Good, 1=Bad)')
plt.suptitle('')
plt.tight_layout()
plt.show()
```

```python
# Default rate by categorical features
cat_cols = ['Sex', 'Housing', 'Saving accounts', 'Checking account', 'Purpose', 'Job']
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, col in zip(axes.flatten(), cat_cols):
    df.groupby(col)['Risk'].mean().sort_values(ascending=False).plot(
        kind='bar', ax=ax, color='steelblue', edgecolor='black')
    ax.set_title(f'Default Rate by {col}')
    ax.set_ylabel('Default Rate')
    ax.tick_params(axis='x', rotation=45)
plt.suptitle('Default Rate by Categorical Features')
plt.tight_layout()
plt.show()
```

```python
# Pairplot of numeric features colored by Risk
sns.pairplot(df[['Age', 'Credit amount', 'Duration', 'Risk']],
             hue='Risk', palette={0: 'steelblue', 1: 'tomato'}, diag_kind='kde')
plt.suptitle('Pairplot of Numeric Features by Risk', y=1.02)
plt.show()
```
